**Lambda Functions:**

These are anonymous, small, and single-expression functions.

```python
square = lambda x: x * x
result = square(5)  # result = 25
```

**Recursive Functions:**

hese functions call themselves within their definition.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

result = factorial(5)  # result = 120
```

**Higher-order Functions:**

These functions take other functions as arguments or return functions as results.

Example: map(), filter(), sorted()

```python
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x * x, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

**Generator Functions:**

These functions use the yield keyword to return an iterator object, generating values on demand.

```python
def even_numbers(n):
    for i in range(n):
        if i % 2 == 0:
            yield i

for number in even_numbers(10):
    print(number)  # Output: 0, 2, 4, 6, 8
```

Higher-order: Functions that take other functions as arguments or return functions.
Generator: Functions that use yield to generate values on demand.

Lesson # 9 **EXCEPTIO N HANDLING**

**try and except**

The try block contains the code that might raise an exception. The except block is executed if an exception of the specified type occurs within the try block.

```python
try:
    result = 10 / 0  # This will raise a ZeroDivisionError
except ZeroDivisionError:
    print("Error: Division by zero")
```

**Multiple except Blocks**

You can have multiple except blocks to handle different types of exceptions.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Division by zero")
except ValueError:
    print("Error: Invalid input")
```

Else BLock

The else block is executed if no exception occurs in the try block.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Division by zero")
except ValueError:
    print("Error: Invalid input")
else:
    print("Result:", result)
```

**finally Block**

The finally block is always executed, regardless of whether an exception occurred or not. It's often used for cleanup operations.

```python
try:
    file = open("myfile.txt", "r")
    # Perform operations with the file
except FileNotFoundError:
    print("Error: File not found")
finally:
    file.close()  # Ensure the file is closed
```

Example: Comprehensive Exception Handling

```python
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    result = num1 / num2
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
except ValueError:
    print("Error: Invalid input. Please enter numbers only")
else:
    print("Result:", result)
finally:
    print("This will always be executed")
```

Explanation

try: Encloses the code that might raise an exception.
except: Catches and handles specific exceptions.
else: Executes if no exception occurs in the try block.
finally: Always executes, typically used for cleanup.
Benefits of Exception Handling

Prevents program crashes due to unexpected errors.
Allows for graceful handling of errors and provides informative messages to the user.
Improves code readability and maintainability.

**The return statement**

The return statement is used within a function to:

Exit the function: When a return statement is encountered, the function immediately stops executing and returns control to the caller.

Return a value (optional): The return statement can optionally specify a value to be returned to the caller. If no value is specified, the function returns None implicitly.

Examples

1. Returning a Value

```python
def add(x, y):
    """This function adds two numbers and returns the sum."""
    sum = x + y
    return sum

result = add(5, 3)
print(result)  # Output: 8
```

2. Returning Multiple Values

You can return multiple values from a function by separating them with commas. They are returned as a tuple.

```python
def get_name_and_age():
    """This function returns a person's name and age."""
    name = "Alice"
    age = 30
    return name, age

name, age = get_name_and_age()
print(name, age)  # Output: Alice 30
```

3. Returning None

If a function doesn't have an explicit return statement, it implicitly returns None.

```python
def greet(name):
    """This function greets the person passed in as a parameter."""
    print(f"Hello, {name}!")

result = greet("Bob")
print(result)  # Output: None
```

4. Using return for Early Exit

You can use return to exit a function prematurely based on a condition.

```python
def is_even(number):
    """This function checks if a number is even."""
    if number % 2 == 0:
        return True
    else:
        return False

result = is_even(7)
print(result)  # Output: False
```

Importance of return

The return statement is crucial for:

Passing data back to the caller: It allows functions to produce results that can be used in other parts of the program.

Controlling function flow: It enables functions to exit early or return different values based on conditions.

Making functions reusable: Functions with clear return values can be easily incorporated into larger programs.

# ***Lesson # 10 File Handling ***

File handling involves working with files on your computer, such as reading data from them, writing data to them, or manipulating their contents. In Python, you can achieve this using built-in functions and the open() function.

Opening a File

The open() function is used to open a file and returns a file object. It takes two main arguments:

File path: The location of the file on your system.
Mode: Specifies how you want to open the file (read, write, append, etc.).

File Modes

'r': Read mode (default). Opens the file for reading.
'w': Write mode. Opens the file for writing, creating it if it doesn't exist, and overwriting its contents if it does.
'a': Append mode. Opens the file for writing, creating it if it doesn't exist, and appending new data to the end.
'x': Create mode. Creates a new file for writing, raising an error if it already exists.
't': Text mode (default). Opens the file in text mode, where data is interpreted as strings.
'b': Binary mode. Opens the file in binary mode, where data is interpreted as bytes.

**Reading from a File**

```python
file = open("myfile.txt", "r")
content = file.read()  # Read the entire file content
print(content)
file.close()
```

**Writing to a File**

```python
file = open("myfile.txt", "w")
file.write("Hello, world!\n")  # Write a string to the file
file.close()
```

**Appending to a File**

```python
file = open("myfile.txt", "a")
file.write("This is another line.\n")  # Append to the file
file.close()
```

Using with Statement

The with statement provides a convenient way to handle files, ensuring they are automatically closed when the block is exited.

```python
with open("myfile.txt", "r") as file:
    content = file.read()
    print(content)
```

Other File Operations

file.readline(): Reads a single line from the file.
file.readlines(): Reads all lines from the file and returns them as a list.
file.tell(): Returns the current position within the file.
file.seek(): Moves the current position to a specific location in the file.

**Example: Working with CSV Files**

```python
import csv

with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)  # Process each row of the CSV data
```

**File Pointer**

In Python, when you open a file using the open() function, you get a file object. This file object has an associated file pointer, which is an internal marker that keeps track of the current position within the file.

How it Works

Initially, the file pointer is positioned at the beginning of the file.
As you read or write data to the file, the file pointer moves forward to indicate the next position for reading or writing.
You can manipulate the file pointer using methods like seek() and tell() to access specific locations within the file.

Methods for File Pointer Manipulation

tell(): This method returns the current position of the file pointer as an integer representing the number of bytes from the beginning of the file.

```python
file = open("myfile.txt", "r")
position = file.tell()
print(position)  # Output: 0 (initially at the beginning)
file.close()
```

seek(): This method allows you to move the file pointer to a specific position within the file. It takes two arguments:

offset: The number of bytes to move the pointer.
whence (optional): Specifies the reference point for the offset. It can be:
0: Beginning of the file (default)
1: Current position
2: End of the file

```python
file = open("myfile.txt", "r")
file.seek(10)  # Move 10 bytes from the beginning
position = file.tell()
print(position)  # Output: 10
file.close()
```

Why File Pointers are Important

Sequential Access: They enable sequential reading or writing of data from the file, ensuring you access information in the desired order.
Random Access: They allow you to jump to specific locations within the file, facilitating efficient data retrieval or modification.
Data Integrity: They help maintain the integrity of the file by ensuring you read or write data from the correct position.

**Lesson # 11** The Math and Date TIme Calendar

Epoch

In the context of date and time, the epoch refers to a specific point in time used as a reference for calculating other dates and times. It's essentially the starting point of a particular timekeeping system.

Unix Epoch

The most widely used epoch is the Unix epoch, which is defined as January 1, 1970, at 00:00:00 Coordinated Universal Time (UTC). Many computer systems and programming languages, including Python, use this epoch as their reference point for representing time.

Representing Time

Time is often represented as the number of seconds (or milliseconds) that have elapsed since the epoch. This value is commonly known as a timestamp or Unix timestamp.

**Example in Python**

```python
import time

epoch_time = time.time()  # Get current Unix timestamp
print(epoch_time)  # Output: Number of seconds since the epoch
```

Converting Timestamp to Human-Readable Date

You can convert a timestamp back to a human-readable date and time using functions like time.localtime() or datetime.datetime.fromtimestamp().

```python
import time

epoch_time = 1678886400  # Example timestamp
local_time = time.localtime(epoch_time)
print(time.strftime("%Y-%m-%d %H:%M:%S", local_time))  # Output: 2023-03-15 00:00:00
```

Importance of Epoch

Reference Point: Provides a consistent and standardized reference point for calculating dates and times.
Time Representation: Allows time to be represented as a single numerical value, which is convenient for storage and manipulation.
Time Comparison: Facilitates easy comparison of dates and times by simply comparing their corresponding timestamps.

**The datetime Module**

Python's datetime module provides classes for manipulating dates and times. The most commonly used classes are:

datetime.date: Represents a date (year, month, day).
datetime.time: Represents a time (hour, minute, second, microsecond).
datetime.datetime: Represents a date and time combined.
datetime.timedelta: Represents the difference between two dates or times.

Commonly Used Methods

Here are some frequently used methods for these classes:

1. date Class:

today(): Returns the current local date.
fromtimestamp(timestamp): Creates a date object from a Unix timestamp.
weekday(): Returns the day of the week as an integer (0 is Monday).
isoweekday(): Returns the day of the week as an integer (1 is Monday).
strftime(format): Formats the date as a string according to the given format codes.
2. time Class:

now(): Returns the current local time.
strftime(format): Formats the time as a string according to the given format codes.
3. datetime Class:

now(): Returns the current local date and time.
combine(date, time): Combines a date and time object into a datetime object.
fromtimestamp(timestamp): Creates a datetime object from a Unix timestamp.
date(): Returns the date part of the datetime object.
time(): Returns the time part of the datetime object.
strftime(format): Formats the datetime as a string according to the given format codes.
4. timedelta Class:

days: The number of days in the timedelta.
seconds: The number of seconds in the timedelta.
microseconds: The number of microseconds in the timedelta.
total_seconds(): Returns the total number of seconds in the timedelta.

```python
import datetime

# Get current date and time
now = datetime.datetime.now()
print(now)

# Format the date and time
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print(formatted_date)

# Create a timedelta
delta = datetime.timedelta(days=5, hours=12)
print(delta)

# Add the timedelta to the current date and time
future_date = now + delta
print(future_date)
```
